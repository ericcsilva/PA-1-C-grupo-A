**Informações importantes!**


*   *1.* Quando o usuário é **menor** de 16 anos o sistema para - break;
*   **2.** Quando o jogador tem **entre** 16 e 17 anos o sistema pede o nome do resposável do usuário e pede um código que é "**PAIS2026**" e também uma senha que é "**1234**";
* **3.** Quando o usuário é **maior** que 18 anos o sistema o libera automaticamente;
* **4.** Ao final de cada cadastro o sistema pergunta se outro usuário vai se cadastrar. Se sim **digite "s"**, caso contrário **digite "n"**.







```python
"""
Login de um jogo para maiores de 18 anos.

Regras de idade:
  - menos de 16 anos  -> acesso negado
  - 16 e 17 anos      -> precisa da autorização dos pais
  - 18 anos ou mais   -> acesso liberado (sem autorização)

Uso de:
  - continue -> volta a pedir o dado quando a entrada está errada
  - break    -> sai do laço (dado correto, acesso negado ou fim do programa)
  - pass     -> caso em que nenhuma ação extra é necessária
"""

SENHA_JOGO = "1234"
CODIGO_PAIS = "PAIS2026"
MAX_TENTATIVAS = 3


def pedir_apelido():
    while True:
        apelido = input("Apelido (nickname): ").strip()

        if not apelido:
            print("  -> O apelido não pode ficar vazio.")
            continue

        if len(apelido) < 3:
            print("  -> O apelido precisa ter pelo menos 3 letras.")
            continue

        break
    return apelido


def pedir_idade():
    while True:
        texto = input("Idade: ").strip()

        if not texto.isdigit():
            print("  -> Digite apenas números inteiros.")
            continue

        idade = int(texto)

        if idade > 120:
            print("  -> Idade impossível. Digite novamente.")
            continue

        break
    return idade


def pedir_autorizacao_dos_pais():
    """Pede o nome do responsável e o código de autorização. Retorna True/False."""
    print("\n  Você tem menos de 18 anos. É necessária a autorização dos pais.")

    while True:
        responsavel = input("  Nome do responsável: ").strip()
        if not responsavel:
            print("  -> Informe o nome do responsável.")
            continue
        break

    tentativas = MAX_TENTATIVAS
    while tentativas > 0:
        codigo = input(f"  Código de autorização ({tentativas} tentativa(s)): ").strip()

        if codigo != CODIGO_PAIS:
            tentativas -= 1
            print("  -> Código incorreto.")
            continue

        print(f"  Autorização confirmada por {responsavel}!")
        return True

    return False


def fazer_login(apelido):
    tentativas = MAX_TENTATIVAS
    while tentativas > 0:
        senha = input(f"Senha ({tentativas} tentativa(s)): ")

        if senha != SENHA_JOGO:
            tentativas -= 1
            print("  -> Senha incorreta.")
            continue

        print(f"\nLogin realizado! Bem-vindo(a) ao jogo, {apelido}!")
        return True

    print("\nMuitas tentativas erradas. Conta bloqueada temporariamente.")
    return False


def jogador_entra():
    """Executa todo o fluxo para um jogador."""
    apelido = pedir_apelido()
    idade = pedir_idade()

    if idade < 16:
        print("\nAcesso NEGADO: o jogo não é permitido para menores de 16 anos.")
        return

    if idade < 18:
        autorizado = pedir_autorizacao_dos_pais()
        if not autorizado:
            print("\nAcesso NEGADO: autorização dos pais não confirmada.")
            return
    else:
        pass  # maior de 18: não precisa de autorização

    fazer_login(apelido)


def main():
    print("=== LOGIN DO JOGO (+18) ===")

    while True:
        print()
        jogador_entra()

        while True:
            outro = input("\nOutro jogador quer entrar? (s/n): ").strip().lower()
            if outro not in ("s", "n"):
                print("  -> Digite 's' ou 'n'.")
                continue
            break

        if outro == "n":
            break

    print("\nServidor encerrado. Até a próxima!")


if __name__ == "__main__":
    main()
```
