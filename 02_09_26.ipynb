4. Soma e média
Crie uma lista contendo 5 notas de um aluno.
Por exemplo:
notas = [7.5, 8.0, 6.5, 9.0, 8.5]
Calcule:
 A soma das notas.
 A média.
 A maior nota.
 A menor nota.
Desafio: determine se o aluno foi aprovado considerando média maior ou igual
a 7.

```python
notas = [7.5, 8.0, 6.5, 9.0, 8.5]
soma = sum(notas)
media = sum(notas) / len(notas)
maior = max(notas)
menor = min(notas)

print(media)
print(soma)
print(maior)
print(menor)

if media >= 7:
  print('aluno aprovado')

else:
  print('aluno reprovado')
```

7. Separando números positivos e negativos
Crie uma lista contendo números positivos, negativos e zero:
numeros = [-5, 10, -3, 8, 0, -1, 7, -9, 4]
Crie duas novas listas:
 Uma contendo apenas números positivos.
 Outra contendo apenas números negativos.
Depois mostre as duas listas.
Desafio: crie uma terceira lista contendo apenas os zeros.

```python
numeros = [-5, 10, -3, 8, 0, -1, 7, -9, 4]

positivos = []
negativos = []
zero = []

for num in numeros:
    if num > 0:
        positivos.append(num)
    elif num == 0:
        zero.append(num)
    else:
        negativos.append(num)

print(f"Números Positivos: {positivos}")
print(f"Números Negativos: {negativos}")
print(f"Zeros: {zero}")
```

8. Ordenando uma lista
Crie uma lista com 10 números escolhidos por você.
Depois:
 Mostre a lista original.
 Ordene a lista em ordem crescente.
 Ordene a lista em ordem decrescente.
 Mostre o maior e o menor valor.
Desafio: faça uma cópia da lista antes de ordená-la, mantendo a lista original
intacta.

```python
numeros = [23, 5, 78, 12, 45, 9, 67, 34, 1, 90]


numeros_copia = numeros.copy()


print("Lista original:", numeros)


numeros_copia.sort()
print("Lista em ordem crescente:", numeros_copia)


numeros_copia.sort(reverse=True)
print("Lista em ordem decrescente:", numeros_copia)

print("Maior valor:", max(numeros))
print("Menor valor:", min(numeros))


print("Lista original (intacta):", numeros)
```

9. Cadastro de produtos
Crie uma lista contendo 5 produtos. Para cada produto, armazene:

 Nome
 Preço
Uma possibilidade é:
produtos = [
["Arroz", 25.90],
["Feijão", 8.50],
["Macarrão", 6.90],
["Leite", 5.50],
["Café", 15.00]
]
Faça um programa que:
 Mostre todos os produtos e seus preços.
 Calcule o valor total da compra.
 Encontre o produto mais caro.
 Encontre o produto mais barato.
Desafio: permita que o usuário informe um produto e verifique se ele está
cadastrado.


```python

produtos = [
    ["Uva", 9.90],
    ["Sabão", 4.50],
    ["Macarrão", 6.90],
    ["Óleo", 8.30],
    ["Shampoo", 18.90]
]


print("=== Lista de Produtos ===")
for produto in produtos:
    nome = produto[0]
    preco = produto[1]
    print(f"{nome}: R$ {preco:.2f}")


total = 0
for produto in produtos:
    total += produto[1]
print(f"\nValor total da compra: R$ {total:.2f}")


mais_caro = produtos[0]
for produto in produtos:
    if produto[1] > mais_caro[1]:
        mais_caro = produto
print(f"Produto mais caro: {mais_caro[0]} (R$ {mais_caro[1]:.2f})")


mais_barato = produtos[0]
for produto in produtos:
    if produto[1] < mais_barato[1]:
        mais_barato = produto
print(f"Produto mais barato: {mais_barato[0]} (R$ {mais_barato[1]:.2f})")


print("\n=== Consulta de Produto ===")
busca = input("Digite o nome de um produto para verificar se está cadastrado: ").lower()

encontrado = False
for produto in produtos:
    if produto[0].lower() == busca.lower():
        encontrado = True
        print(f"Produto encontrado! {produto[0]} custa R$ {produto[1]:.2f}")
        break

if not encontrado:
    print(f"O produto '{busca}' não está cadastrado.")
```

10. Desafio final — sistema de notas
Crie um programa que permita cadastrar as notas de vários alunos em uma lista.
O programa deve:
Perguntar quantos alunos serão cadastrados.
Para cada aluno, pedir:
o Nome
o Nota
Armazenar as informações em uma lista.
Ao final, mostrar:
o Todos os alunos e suas notas.
o A média da turma.
o O aluno com a maior nota.
o O aluno com a menor nota.
o Quais alunos foram aprovados (nota >= 7).
o Quais alunos foram reprovados (nota < 7).
Desafio extra: permita que o usuário pesquise um aluno pelo nome e veja sua
nota.

```python

alunos = []


quantidade = int(input("Quantos alunos serão cadastrados? "))


for i in range(quantidade):
    print(f"\n--- Aluno {i + 1} ---")
    nome = input("Nome: ")
    nota = float(input("Nota: "))
    nota1 = float(input("Nota: "))
    alunos.append([nome, nota])


print("\n=== Lista de Alunos ===")
for aluno in alunos:
    print(f"{aluno[0]}: {aluno[1]:.1f}")


soma_notas = 0
for aluno in alunos:
    soma_notas += aluno[1]
media = soma_notas / len(alunos)
print(f"\nMédia da turma: {media:.2f}")


maior_nota = alunos[0]
for aluno in alunos:
    if aluno[1] > maior_nota[1]:
        maior_nota = aluno
print(f"Aluno com a maior nota: {maior_nota[0]} ({maior_nota[1]:.1f})")


menor_nota = alunos[0]
for aluno in alunos:
    if aluno[1] < menor_nota[1]:
        menor_nota = aluno
print(f"Aluno com a menor nota: {menor_nota[0]} ({menor_nota[1]:.1f})")


aprovados = []
reprovados = []
for aluno in alunos:
    if aluno[1] >= 7:
        aprovados.append(aluno)
    else:
        reprovados.append(aluno)

print("\n=== Aprovados (nota >= 7) ===")
if aprovados:
    for aluno in aprovados:
        print(f"{aluno[0]}: {aluno[1]:.1f}")
else:
    print("Nenhum aluno aprovado.")

print("\n=== Reprovados (nota < 7) ===")
if reprovados:
    for aluno in reprovados:
        print(f"{aluno[0]}: {aluno[1]:.1f}")
else:
    print("Nenhum aluno reprovado.")


print("\n=== Pesquisar Aluno ===")
busca = input("Digite o nome do aluno que deseja pesquisar: ")

encontrado = False
for aluno in alunos:
    if aluno[0].lower() == busca.lower():
        encontrado = True
        print(f"{aluno[0]} tem nota {aluno[1]:.1f}")
        break

if not encontrado:
    print(f"Aluno '{busca}' não encontrado.")
```
